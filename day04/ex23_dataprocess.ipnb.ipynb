```python
# matplotlib 라이브러리 로드
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]      #x축 데이터
y=[10,20,15,30,20]   #y축 데이터

plt.plot(x,y)        #라인차트
plt.show()           #필요없는 차트설명 제외
```

제목,레이블 표시
    -기본적으로 한글 출력이 안됨

```python
plt.plot(x,y)
plt.title('점수 추이')
plt.xlabel('시간')
plt.ylabel('score')
plt.show
```

-한글설정


```python
# 리소스 인수, 폰트관리자, 리소스 포함
from matplotlib import rcParams,font_manager,rc

font_path = 'C:/Windows/Fonts/malgun.ttf'  #NanumGothic.ttf
font = font_manager.FontProperties(fname=font_path).get_name()
rc('font',family=font)
rcParams['axes.unicode_minus']=False # 마이너스 표시가 깨짐방지
```

```python
plt.plot(x,y)
plt.title('점수 추이')      # 차트 제목
plt.xlabel('시간')          # x축 라벨
plt.ylabel('score')         # y축 라벨
plt.show
```

```python
plt.plot(x,y,marker='s')    #마커표시 *,^,
plt.title('점수 추이')      # 차트 제목
plt.xlabel('시간')          # x축 라벨
plt.ylabel('score')         # y축 라벨
plt.show
```

#### 라인,색상,포인트
-x,y 좌표 뒤에 문자열로 작성
    -색상:'b,g,r,c,m.y,k,w'
    -마커: 위와 동일
    -라인스타일-(line),--(dashed line),-(dashed dot line),:(dotted line)


```python
plt.plot(x,y,'r^--')
plt.title('점수 추이')
plt.xlabel('시간')
plt.ylabel('Score')
plt.show()
```

#### 스타일지정

-seaborn은 차트 생성전에 스타일 지정


```python
sns.set_style('whitegrid')#한글깨짐 방지
sns.set_theme(font=font,rc={'axes.unicode_minus':False})#한글깨짐 방지
```

```python
#막대 그래프
plt.bar(x,y)
plt.title('점수추이')
```

#### 차트종류
-plot() : 라인차트
-bar() : 막대그래프
-barh() : 가로막대그래프.x,y축 값을 변경
-hist() : 히스토그램. 분포표
-scatter() : 산점도. 데이터간의 관계 표현
-pie() : 파이차트. 비율표시

#### 차트 팁
-grid(True) : 맷 플롯에서만 영향미침. seaborn에서는 영향없음

```python
# 산점도
study_time = [1,2,3,4,5,6,7,8]
score = [50,55,65,72,78,81,93,98]

plt.scatter(x=study_time,y=score)
plt.title('학습시간 대비 점수 상관관계')
```

```python
# 주피터노트북 새로 생성한 뒤 첫번째 코드 라이브러리 배치

import matplotlib.pyplot as plt
import seaborn as sns

import numpy as np
```

```python
x=[1,2,3,4]
y=[10,20,15,30]
z=[5,10,8,12]

fig = plt.figure(figsize=(10,5)) #새로운 차트를 선언해서 변수할당
ax=fig.add_subplot(111,projection='3d')

ax.scatter(x,y,z)

ax.set_xlabel('x축')
ax.set_ylabel('x축')
ax.set_zlabel('x축')
```

```python
#3D Space
x = np.linspace(-5,5,50) #-5~5사이를 50개 간격으로 쪼갬
y = np.linspace(-5,5,50) #-5~5사이를 50개 간격으로 쪼갬

x,y = np.meshgrid(x,y)
z=x**2+y**2

fig =plt.figure()
ax =fig.add_subplot(111,projection='3d')
ax.plot_surface(x,y,z)

plt.show()
```

## 기본 사용


````python
# wordcloud 라이브러리 코드
from wordcloud import WordCloud
import matplotlib.pyplot as plt
````

````python
text='''python is my favorite programming language data analysis.python ai ai ai
        deep learning, machine learning, contemorary ai skill
     '''
wc = WordCloud().generate(text)
````

````python
plt.imshow(wc)
plt.axis('off')
plt.show()
````

````python
text = '데이터 분석 파이썬 인공지능 인공지능 인공지응 데이터 파이썬 인공지능 c++ java 인공지능 c#'

wc =WordCloud(
    font_path='C:/Windows/Fonts/NanumGothicExtraBold.ttf',
    background_color='azure',
    colormap='cool').generate(text)



plt.imshow(wc)
plt.axis('off')
plt.show()
````
